## Load a Waymo frame


```python
frames = sorted(glob.glob('/content/data/night/front_images_night/*.jpg'))
frame_path = frames[100]

frame_bgr = cv2.imread(frame_path)
frame_rgb = cv2.cvtColor(frame_bgr, cv2.COLOR_BGR2RGB)

preprocess = T.Compose([
    T.Resize((H, W)),
    T.ToTensor(),
    T.Normalize(mean=[0.485, 0.456, 0.406], std=[0.229, 0.224, 0.225]),
])
x    = preprocess(Image.fromarray(frame_rgb)).unsqueeze(0)   # (1,3,H,W) normalized
x_np = x.numpy()                                             # numpy for ONNX Runtime

plt.imshow(cv2.resize(frame_rgb, (W, H))); plt.axis('off')
plt.title('Input: Waymo driving frame'); plt.show()
```

```python
COLORS = np.array([
    [240,  40,  40],   # 0 background
    [180,  60, 200],   # 1 foreground (cars / pedestrians)
    [ 80, 200,  80],   # 2 drivable road
], dtype=np.uint8)


from tqdm.auto import tqdm

def bench(fn, label, n=10):
    """Warmup + n timed runs. Shows a progress bar so it doesn't look frozen."""
    fn()
    times = []
    for _ in tqdm(range(n), desc=label, leave=False):
        t0 = time.perf_counter()
        fn()
        times.append((time.perf_counter() - t0) * 1000)
    times = np.array(times)
    print(f"{label:20}  {times.mean():6.1f} ms  ({1000/times.mean():5.1f} FPS)")
    return times.mean()
```

## Part 1 — Run as PyTorch

Load the network class, load the trained weights into it, run a frame through, and **time it**. The latency number is the baseline we'll compare against in Part 3.


```python
model = SceneSegNetwork()
state_dict = torch.load('/content/models/SceneSeg.pth', map_location='cpu', weights_only=True)
model.load_state_dict(state_dict)
model.eval()

with torch.no_grad():
    pt_logits = model(x).numpy()
pt_map = np.argmax(pt_logits[0], axis=0).astype(np.uint8)

print(f"Output shape: {pt_logits.shape}\n")

# Visual
input_img = cv2.resize(frame_rgb, (W, H))
overlay   = cv2.addWeighted(input_img, 0.5, COLORS[pt_map], 0.5, 0)

fig, axes = plt.subplots(1, 3, figsize=(18, 4))
axes[0].imshow(input_img);     axes[0].set_title('Input frame');           axes[0].axis('off')
axes[1].imshow(COLORS[pt_map]); axes[1].set_title('PyTorch — segmentation'); axes[1].axis('off')
axes[2].imshow(overlay);       axes[2].set_title('PyTorch — overlay');     axes[2].axis('off')
plt.tight_layout(); plt.show()

# Benchmark
def run_pt():
    with torch.no_grad():
        return model(x)
pt_ms = bench(run_pt, 'PyTorch (CPU)')
```

### Heads up — ONNX export is a moving target

If you're watching this video weeks or months after I recorded it, **the export above might fail on your machine first try**. Not because anything's wrong with you — because ONNX export is the most fragile step in any deployment pipeline. Two things move under your feet constantly:

#### 1. Opset versions

ONNX has versioned "operator sets" — opset 17, 18, 19... Each version changes how some ops are spelled. The exporter wants the latest, the converter to lower opsets sometimes can't bridge the gap. The PyTorch ↔ ONNX ↔ TensorRT versions all have to line up.

Symptoms you might see:
- `Failed to convert the model to the target version 17 using the ONNX C API`
- `No initializer or constant input to node found`
- A 0.7 MB file that should have been 30 MB

Fix path:
- **Try bumping `opset_version` by 1.** We're at 18 here; if 18 breaks, try 19.
- If TensorRT downstream complains about the opset being too new, lower it instead.
- The highest opset your TensorRT version supports is your ceiling.

#### 2. `dynamo=True` vs `dynamo=False`

PyTorch 2.5+ ships a new ONNX exporter (`dynamo=True`) that uses `torch.export`. It's smarter on attention-heavy models but stricter — it catches model shapes the legacy tracer would silently smooth over.

Common failure modes:
- `Exporting a ScriptModule is not supported` → dynamo doesn't accept TorchScript-traced models. Use `dynamo=False`, or load from `.pth` + class definition (what we did here).
- `tried to trace X but it is not part of the active trace` → you wrapped a ScriptModule inside a regular `nn.Module`. Same fix.
- Random shape errors → check for hardcoded `reshape([...])` calls that bake in batch=1 (we hit and fixed exactly this in `SceneSeg.py`).

Pragmatic rule:
- **Plain CNN** like SceneSeg → `dynamo=False`, simpler, equivalent ONNX.
- **Transformer / attention-heavy** → try `dynamo=True` first (see `Bonus_ONNX_Dynamo.ipynb`).
- **It still fails?** That's a bug worth reporting on PyTorch's issue tracker. The ONNX exporter team is responsive on real bugs.

#### TL;DR

If this notebook fails for you, the fix is almost certainly **one flag away** — `opset_version` or `dynamo`. Don't assume your model is broken; assume the export pipeline drifted under it. Welcome to deployment.

## Bonus — What's inside the file?

ONNX files are graphs: nodes (operators) connected by tensors. Quick look:


```python
m = onnx.load('/content/models/SceneSeg.onnx')

op_counts = {}
for node in m.graph.node:
    op_counts[node.op_type] = op_counts.get(node.op_type, 0) + 1

total_ops  = sum(op_counts.values())
unique_ops = len(op_counts)
top_5      = sorted(op_counts.items(), key=lambda kv: -kv[1])[:5]

print(f"Total nodes:       {total_ops}")
print(f"Unique op types:   {unique_ops}")
print(f"Inputs / outputs:  {len(m.graph.input)} / {len(m.graph.output)}")
print(f"\nTop 5 operators:")
for op, n in top_5:
    print(f"  {op:<20} {n}")

print(f"\nTip: open SceneSeg.onnx in https://netron.app for the visual graph.")
```
